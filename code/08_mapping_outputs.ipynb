## Note on data inputs

This notebook documents the **analysis method**. The raw licensed-facility inputs it reads (California Community Care Licensing / CDPH files containing facility administrator names, telephone numbers, and complaint histories) are **not distributed** with this repository, to protect vulnerable populations. De-identified and aggregated outputs are included; see `docs/DATA_AVAILABILITY.md`.

_As a result, this notebook is provided for transparency of method and will not run end-to-end against this public repository alone._

```python
eldercare_map = eldercare.rename(columns={
    "Facility Name": "institution_name",
    "* Damage": "damage_class",
    "Facility Capacity": "capacity"
}).copy()

eldercare_map["category"] = "Elder care"
eldercare_map["source_file"] = "ResidentialElderCareFacility05172026.csv"

eldercare_map = eldercare_map[
    [
        "institution_name",
        "category",
        "damage_class",
        "capacity",
        "latitude",
        "longitude",
        "source_file",
        "geometry"
    ]
]
```

```python
adult_res_map = adult_res.rename(columns={
    "Facility Name": "institution_name",
    "* Damage": "damage_class",
    "Facility Capacity": "capacity"
}).copy()

adult_res_map["category"] = "Adult residential/group home"
adult_res_map["source_file"] = "AdultResidentialFacilities05172026.csv"

adult_res_map = adult_res_map[
    [
        "institution_name",
        "category",
        "damage_class",
        "capacity",
        "latitude",
        "longitude",
        "source_file",
        "geometry"
    ]
]
```

```python
all_social_infra = pd.concat(
    [
        eldercare_map,
        adult_res_map
        # add childcare_map, healthcare_map, schools_map, churches_map later
    ],
    ignore_index=True
)

all_social_infra = gpd.GeoDataFrame(
    all_social_infra,
    geometry="geometry",
    crs="EPSG:4326"
)
```
